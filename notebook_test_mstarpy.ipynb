```python
import string
import mstarpy as ms

seeds = list(string.ascii_lowercase) + [str(i) for i in range(10)]
all_rows = {}

for seed in seeds:
    try:
        rows = ms.screener_universe(
            seed,
            field=["name", "isin", "sector"],
            filters={"investmentType": "EQ"}
        )
        for r in rows:
            meta = r.get("meta", {})
            if meta.get("exchange") == "XKLS":
                key = meta.get("securityID") or meta.get("ticker")
                if key:
                    all_rows[key] = r
    except Exception:
        pass

malaysia_stocks = list(all_rows.values())
print("unique Malaysia stocks:", len(malaysia_stocks))
```
